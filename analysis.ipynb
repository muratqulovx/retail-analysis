```python
import pandas as pd

def simplify_categories_20(df, category_col='Category', type_col='Type'):
    # Mapping 20 kategoriya uchun
    category_mapping = {
        'dessert': 'Dessert',
        'Dessert': 'Dessert',
        'drink': 'Drink',
        'dairy': 'Dairy',
        'bakery': 'Bakery',
        'grain_pasta': 'Grains & Pasta',
        'beans_legumes': 'Grains & Pasta',
        'nut': 'Grains & Pasta',
        'Snack': 'Snack',
        'snack': 'Snack',
        'fruit_vegetable': 'Fruit & Vegetable',
        'Fruit_Vegetable': 'Fruit & Vegetable',
        'meat_fish_poultry': 'Meat & Fish',
        'condiment': 'Pantry',
        'Condiment': 'Pantry',
        'canned_frozen_dried': 'Pantry',
        'ingredients': 'Pantry',
        'baby food': 'Baby',
        'baby_care': 'Baby',
        'Baby_Care': 'Baby',
        'personal_care': 'Personal Care',
        'Personal_Care': 'Personal Care',
        'household': 'Household',
        'Household': 'Household',
        'health_medical': 'Health & Medical',
        'stationery_office': 'Stationery & Office',
        'clothing_accessories': 'Clothing',
        'media': 'Media & Electronics',
        'electronics_accessories': 'Media & Electronics',
        'tools_hardware': 'Tools & Home',
        'home_decor': 'Tools & Home',
        'packaging': 'Packaging',
        'Other Food': 'Other Food',
        'Other Non-Food': 'Other Non-Food',
        'other_non_food': 'Other Non-Food',
        'Unknown': 'Unknown'
    }

    # Food toifalari
    food_categories = {
        'Dessert', 'Drink', 'Dairy', 'Bakery', 'Grains & Pasta', 'Snack',
        'Fruit & Vegetable', 'Meat & Fish', 'Pantry', 'Baby', 'Other Food'
    }

    # Kategoriya soddalashtirish
    df[category_col] = df[category_col].map(category_mapping).fillna('Other Non-Food')

    # Type aniqlash
    def assign_type(cat):
        if cat == 'Unknown':
            return 'Unknown'
        elif cat in food_categories:
            return 'Food'
        else:
            return 'Non-Food'

    df[type_col] = df[category_col].apply(assign_type)
    return df
merged_df = simplify_categories_20(merged_df, category_col='Category', type_col='Type')
```
