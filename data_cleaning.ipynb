### Russian to English

```python
column_mapping = {
    'Чек': 'Check',
    'Номенклатура': 'Product Name',
    'Количество сторно': 'Cancelled Quantity',
    'Количество': 'Quantity',
    'Сумма сторно': 'Cancelled Amount',
    'Сумма': 'Amount',
    'Сумма скидки': 'Discount Amount',
    'Количество позиций сторно': 'Cancelled Item Types',
    'Количество позиций': 'Item Types',
    'Количество чеков': 'Number of Checks',
    'Сумма сторно.1': 'Cancelled Amount (Checks)',
    'Сумма.1': 'Total Amount (Checks)',
    'Сумма скидки.1': 'Discount Amount (Checks)',
    'Время на чек': 'Time per Check'
}

data.rename(columns=column_mapping, inplace=True)
data.fillna(0, inplace=True)
```

```python
data["code"] = None

code = data.loc[0, "Check"]
data.loc[0, "code"] = code

for i in range(len(data)):
    if "Чек" in str(data.loc[i, "Check"]):
        data.loc[i, "code"] = code
    else:
        code = data.loc[i, "Check"]
        data.loc[i, "code"] = code

```

```python
def clean_number(x):
    if isinstance(x, str):
        x = x.replace('\xa0', '') 
        x = x.replace(',', '.')    
        x = x.strip()
        try:
            return float(x)
        except ValueError:
            return None  
    return x 

data["Quantity"] = data["Quantity"].apply(clean_number)
data["Cancelled Quantity"] = data["Cancelled Quantity"].apply(clean_number)
data["code"] = data["code"].apply(clean_number)

data["Net Sold"] = data["Quantity"] - data["Cancelled Quantity"]
```
